# Instructions
0. See README.md for instructions on python version and dependencies
1. Connect phone and camera via USB with DTL067
2. Start the SpinView imaging software for the camera
    - to enhance gain of the camera, go to the 'features' subwindow, search for 'gamma' and turn on 'gamma enabled' and increase 'gamma' to max 4.
3. Initiate a Mapping() object
5. Make sure the phone and camera a properly aligned and focussed 
    - Mirror the phone screen
    - Adjust the FOV if necessary by moving the camera lens and camera. A 200mm lens can be used for the largest FOV, see setup_named.png
    - Make sure the project lines are nicely parrallel and orthogonal to the kid array
        - To adjust the rotation of the phone: loosen the cage mount that holds the lens tube such that it can rotate  
9. Input frequency sweep parameters Mapping().add_fsweep(...)
10. Start mapping with Mapping().start()
    - Choose linewidth, color and stepsize of the lines
    - Align the lines and add scans with L
    - Start measurement with Enter
    - Choose the directory to save the data in
    - First a dark sweep is done and plotted
    - Then the rest of the scans is done
9. Shut down phone if desired with Mapping().reboot()
